```python
import os
import random
import shutil

RAW_DIR = "/content/yellow_rust_19/RAW/RAW"
SPLIT_DIR = "/content/yellow_rust_severity"

CLASSES = ["0", "R", "MR", "MRMS", "MS", "S"]

random.seed(42)

# Create split directories
for split in ["train", "val", "test"]:
    for cls in CLASSES:
        os.makedirs(
            os.path.join(SPLIT_DIR, split, cls),
            exist_ok=True
        )

# Split ratios
TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

for cls in CLASSES:

    source_dir = os.path.join(RAW_DIR, cls)

    images = [
        f for f in os.listdir(source_dir)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(images)

    n = len(images)

    train_end = int(n * TRAIN_RATIO)
    val_end = train_end + int(n * VAL_RATIO)

    train_images = images[:train_end]
    val_images = images[train_end:val_end]
    test_images = images[val_end:]

    print(
        f"{cls}: "
        f"Train={len(train_images)}, "
        f"Val={len(val_images)}, "
        f"Test={len(test_images)}"
    )

    for filename in train_images:
        shutil.copy2(
            os.path.join(source_dir, filename),
            os.path.join(SPLIT_DIR, "train", cls, filename)
        )

    for filename in val_images:
        shutil.copy2(
            os.path.join(source_dir, filename),
            os.path.join(SPLIT_DIR, "val", cls, filename)
        )

    for filename in test_images:
        shutil.copy2(
            os.path.join(source_dir, filename),
            os.path.join(SPLIT_DIR, "test", cls, filename)
        )

print("\nDataset split completed.")
print("Location:", SPLIT_DIR)
```

```python
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models
from torchvision.models import ResNet18_Weights

# ============================================================
# Configuration
# ============================================================

DATA_DIR = "/content/yellow_rust_severity"

BATCH_SIZE = 32
NUM_CLASSES = 6

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", DEVICE)

# ============================================================
# Image preprocessing
# ============================================================

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# ============================================================
# Load datasets
# ============================================================

train_dataset = datasets.ImageFolder(
    f"{DATA_DIR}/train",
    transform=train_transform
)

val_dataset = datasets.ImageFolder(
    f"{DATA_DIR}/val",
    transform=test_transform
)

test_dataset = datasets.ImageFolder(
    f"{DATA_DIR}/test",
    transform=test_transform
)

print("\nClasses:")
print(train_dataset.class_to_idx)

print("\nDataset sizes:")
print("Train:", len(train_dataset))
print("Val  :", len(val_dataset))
print("Test :", len(test_dataset))

# ============================================================
# DataLoaders
# ============================================================

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

# ============================================================
# Load pretrained ResNet-18
# ============================================================

weights = ResNet18_Weights.DEFAULT

model = models.resnet18(weights=weights)

# Replace final classification layer
model.fc = nn.Linear(
    model.fc.in_features,
    NUM_CLASSES
)

model = model.to(DEVICE)

print("\nModel:")
print("ResNet-18")
print("Output classes:", NUM_CLASSES)
print("Device:", DEVICE)
```

```python
import torch
import numpy as np
from collections import Counter

# Get training labels
train_labels = [label for _, label in train_dataset.samples]

# Count samples per class
class_counts = Counter(train_labels)

print("Class counts:")
for class_idx, class_name in enumerate(train_dataset.classes):
    print(
        f"{class_name}: "
        f"{class_counts[class_idx]}"
    )

# Calculate balanced class weights
total_samples = len(train_labels)
num_classes = len(train_dataset.classes)

class_weights = []

for class_idx in range(num_classes):
    weight = total_samples / (
        num_classes * class_counts[class_idx]
    )
    class_weights.append(weight)

class_weights = torch.tensor(
    class_weights,
    dtype=torch.float32
).to(DEVICE)

print("\nClass weights:")

for class_idx, class_name in enumerate(train_dataset.classes):
    print(
        f"{class_name}: "
        f"{class_weights[class_idx].item():.4f}"
    )
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# ============================================================
# Loss function with class weights
# ============================================================

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)

# ============================================================
# Optimizer
# ============================================================

optimizer = optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

# ============================================================
# Training configuration
# ============================================================

NUM_EPOCHS = 10

best_val_accuracy = 0.0
best_model_path = "/content/best_yellow_rust_resnet18.pth"

print("Training configuration")
print("----------------------")
print("Model       : ResNet-18")
print("Classes     :", NUM_CLASSES)
print("Epochs      :", NUM_EPOCHS)
print("Learning rate:", 1e-4)
print("Optimizer   : AdamW")
print("Loss        : Weighted CrossEntropyLoss")
print("Device      :", DEVICE)
print("Best model  :", best_model_path)
```

```python
from sklearn.metrics import accuracy_score

# ============================================================
# Training history
# ============================================================

train_losses = []
val_losses = []
train_accuracies = []
val_accuracies = []

# ============================================================
# Training loop
# ============================================================

for epoch in range(NUM_EPOCHS):

    # -------------------------
    # TRAIN
    # -------------------------
    model.train()

    running_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader:

        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        # Clear previous gradients
        optimizer.zero_grad()

        # Forward pass
        outputs = model(images)

        # Calculate loss
        loss = criterion(outputs, labels)

        # Backpropagation
        loss.backward()

        # Update weights
        optimizer.step()

        # Statistics
        running_loss += loss.item() * images.size(0)

        predictions = torch.argmax(outputs, dim=1)

        train_correct += (predictions == labels).sum().item()
        train_total += labels.size(0)

    epoch_train_loss = running_loss / train_total
    epoch_train_accuracy = train_correct / train_total

    # -------------------------
    # VALIDATION
    # -------------------------
    model.eval()

    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            outputs = model(images)

            loss = criterion(outputs, labels)

            val_running_loss += loss.item() * images.size(0)

            predictions = torch.argmax(outputs, dim=1)

            val_correct += (predictions == labels).sum().item()
            val_total += labels.size(0)

    epoch_val_loss = val_running_loss / val_total
    epoch_val_accuracy = val_correct / val_total

    # -------------------------
    # Save history
    # -------------------------

    train_losses.append(epoch_train_loss)
    val_losses.append(epoch_val_loss)

    train_accuracies.append(epoch_train_accuracy)
    val_accuracies.append(epoch_val_accuracy)

    # -------------------------
    # Save best model
    # -------------------------

    if epoch_val_accuracy > best_val_accuracy:

        best_val_accuracy = epoch_val_accuracy

        torch.save(
            model.state_dict(),
            best_model_path
        )

        saved = " ← BEST MODEL SAVED"
    else:
        saved = ""

    # -------------------------
    # Print results
    # -------------------------

    print(
        f"Epoch [{epoch+1}/{NUM_EPOCHS}] "
        f"| Train Loss: {epoch_train_loss:.4f} "
        f"| Train Acc: {epoch_train_accuracy:.4f} "
        f"| Val Loss: {epoch_val_loss:.4f} "
        f"| Val Acc: {epoch_val_accuracy:.4f}"
        f"{saved}"
    )

print("\n" + "=" * 60)
print("TRAINING COMPLETE")
print("=" * 60)

print(f"Best validation accuracy: {best_val_accuracy:.4f}")
print(f"Best model saved at: {best_model_path}")
```

```python
import torch
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# Load BEST validation model
# ============================================================

model.load_state_dict(
    torch.load(
        best_model_path,
        map_location=DEVICE
    )
)

model.eval()

all_labels = []
all_predictions = []
all_probabilities = []

# ============================================================
# Test evaluation
# ============================================================

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(DEVICE)
        labels = labels.to(DEVICE)

        outputs = model(images)

        probabilities = torch.softmax(outputs, dim=1)
        predictions = torch.argmax(outputs, dim=1)

        all_labels.extend(labels.cpu().numpy())
        all_predictions.extend(predictions.cpu().numpy())
        all_probabilities.extend(probabilities.cpu().numpy())

# Convert to NumPy
y_true = np.array(all_labels)
y_pred = np.array(all_predictions)
y_prob = np.array(all_probabilities)

# ============================================================
# Metrics
# ============================================================

accuracy = accuracy_score(y_true, y_pred)

precision_macro = precision_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

recall_macro = recall_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

f1_macro = f1_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

f1_weighted = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

# Multiclass ROC-AUC
roc_auc = roc_auc_score(
    y_true,
    y_prob,
    multi_class="ovr",
    average="macro"
)

# ============================================================
# Print results
# ============================================================

print("=" * 60)
print("YELLOW-RUST-19 RESNET-18 TEST RESULTS")
print("=" * 60)

print(f"Accuracy       : {accuracy:.4f}")
print(f"Macro Precision: {precision_macro:.4f}")
print(f"Macro Recall   : {recall_macro:.4f}")
print(f"Macro F1       : {f1_macro:.4f}")
print(f"Weighted F1    : {f1_weighted:.4f}")
print(f"Macro ROC-AUC  : {roc_auc:.4f}")

# ============================================================
# Per-class report
# ============================================================

print("\n" + "=" * 60)
print("PER-CLASS CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_true,
        y_pred,
        target_names=train_dataset.classes,
        zero_division=0
    )
)

# ============================================================
# Confusion matrix
# ============================================================

cm = confusion_matrix(y_true, y_pred)

print("\n" + "=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

print("Classes:", train_dataset.classes)
print(cm)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models
from torchvision.models import ResNet18_Weights

# ============================================================
# EXPERIMENT 2
# Improved ResNet-18
# ============================================================

# Start again from pretrained ImageNet weights
model_v2 = models.resnet18(
    weights=ResNet18_Weights.DEFAULT
)

# Replace classifier with dropout + 6-class output
model_v2.fc = nn.Sequential(
    nn.Dropout(p=0.30),
    nn.Linear(
        model_v2.fc.in_features,
        NUM_CLASSES
    )
)

model_v2 = model_v2.to(DEVICE)

# ============================================================
# Weighted loss
# ============================================================

criterion_v2 = nn.CrossEntropyLoss(
    weight=class_weights
)

# ============================================================
# Different learning rates
# ============================================================

optimizer_v2 = optim.AdamW(
    [
        {
            "params": model_v2.layer1.parameters(),
            "lr": 1e-5
        },
        {
            "params": model_v2.layer2.parameters(),
            "lr": 1e-5
        },
        {
            "params": model_v2.layer3.parameters(),
            "lr": 1e-5
        },
        {
            "params": model_v2.layer4.parameters(),
            "lr": 1e-5
        },
        {
            "params": model_v2.fc.parameters(),
            "lr": 1e-4
        }
    ],
    weight_decay=1e-4
)

# ============================================================
# Validation-driven LR scheduler
# ============================================================

scheduler_v2 = optim.lr_scheduler.ReduceLROnPlateau(
    optimizer_v2,
    mode="min",
    factor=0.3,
    patience=2,
    min_lr=1e-7
)

# ============================================================
# Configuration
# ============================================================

NUM_EPOCHS_V2 = 15
EARLY_STOPPING_PATIENCE = 4

best_val_f1 = 0.0
epochs_without_improvement = 0

best_model_path_v2 = (
    "/content/best_yellow_rust_resnet18_v2.pth"
)

print("Experiment 2 configuration")
print("=" * 60)
print("Model              : ResNet-18")
print("Pretrained         : ImageNet")
print("Dropout            : 0.30")
print("Loss               : Weighted CrossEntropy")
print("Optimizer          : AdamW")
print("Backbone LR        : 1e-5")
print("Classifier LR      : 1e-4")
print("Scheduler           : ReduceLROnPlateau")
print("Maximum epochs     :", NUM_EPOCHS_V2)
print("Early stopping     :", EARLY_STOPPING_PATIENCE)
print("Selection metric   : Validation Macro-F1")
print("Device             :", DEVICE)
```

```python
from sklearn.metrics import f1_score
import numpy as np

# ============================================================
# Experiment 2 training history
# ============================================================

train_losses_v2 = []
val_losses_v2 = []

train_accs_v2 = []
val_accs_v2 = []

val_f1s_v2 = []
learning_rates_v2 = []

# ============================================================
# Training
# ============================================================

for epoch in range(NUM_EPOCHS_V2):

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    model_v2.train()

    running_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader:

        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        optimizer_v2.zero_grad()

        outputs = model_v2(images)

        loss = criterion_v2(outputs, labels)

        loss.backward()

        optimizer_v2.step()

        running_loss += loss.item() * images.size(0)

        predictions = torch.argmax(outputs, dim=1)

        train_correct += (
            predictions == labels
        ).sum().item()

        train_total += labels.size(0)

    train_loss = running_loss / train_total
    train_accuracy = train_correct / train_total

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    model_v2.eval()

    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    val_true = []
    val_pred = []

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            outputs = model_v2(images)

            loss = criterion_v2(outputs, labels)

            val_running_loss += (
                loss.item() * images.size(0)
            )

            predictions = torch.argmax(
                outputs,
                dim=1
            )

            val_correct += (
                predictions == labels
            ).sum().item()

            val_total += labels.size(0)

            val_true.extend(
                labels.cpu().numpy()
            )

            val_pred.extend(
                predictions.cpu().numpy()
            )

    val_loss = val_running_loss / val_total
    val_accuracy = val_correct / val_total

    val_macro_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0
    )

    # --------------------------------------------------------
    # Learning-rate scheduler
    # --------------------------------------------------------

    scheduler_v2.step(val_loss)

    current_lr = optimizer_v2.param_groups[-1]["lr"]

    # --------------------------------------------------------
    # Save history
    # --------------------------------------------------------

    train_losses_v2.append(train_loss)
    val_losses_v2.append(val_loss)

    train_accs_v2.append(train_accuracy)
    val_accs_v2.append(val_accuracy)

    val_f1s_v2.append(val_macro_f1)
    learning_rates_v2.append(current_lr)

    # --------------------------------------------------------
    # Save best model based on Macro-F1
    # --------------------------------------------------------

    if val_macro_f1 > best_val_f1:

        best_val_f1 = val_macro_f1
        epochs_without_improvement = 0

        torch.save(
            model_v2.state_dict(),
            best_model_path_v2
        )

        saved = " ← BEST MODEL SAVED"

    else:

        epochs_without_improvement += 1
        saved = ""

    # --------------------------------------------------------
    # Print
    # --------------------------------------------------------

    print(
        f"Epoch [{epoch+1}/{NUM_EPOCHS_V2}] "
        f"| Train Loss: {train_loss:.4f} "
        f"| Train Acc: {train_accuracy:.4f} "
        f"| Val Loss: {val_loss:.4f} "
        f"| Val Acc: {val_accuracy:.4f} "
        f"| Val Macro-F1: {val_macro_f1:.4f} "
        f"| LR: {current_lr:.2e}"
        f"{saved}"
    )

    # --------------------------------------------------------
    # Early stopping
    # --------------------------------------------------------

    if epochs_without_improvement >= EARLY_STOPPING_PATIENCE:

        print(
            f"\nEarly stopping triggered after "
            f"{epoch+1} epochs."
        )

        break


print("\n" + "=" * 70)
print("EXPERIMENT 2 TRAINING COMPLETE")
print("=" * 70)

print(f"Best validation Macro-F1: {best_val_f1:.4f}")
print(f"Best model saved at: {best_model_path_v2}")
```

```python
# ============================================================
# IMPROVED ViT-B/16 - YELLOW RUST SEVERITY
# ============================================================

import os
import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import DataLoader
from torchvision import datasets, transforms, models

from sklearn.metrics import f1_score


# ============================================================
# 1. DEVICE
# ============================================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))


# ============================================================
# 2. DATASET PATHS
# ============================================================

DATA_DIR = "/content/yellow_rust_severity"

TRAIN_DIR = os.path.join(DATA_DIR, "train")
VAL_DIR = os.path.join(DATA_DIR, "val")
TEST_DIR = os.path.join(DATA_DIR, "test")


# ============================================================
# 3. TRANSFORMS
# ============================================================

vit_train_transform = transforms.Compose([
    transforms.Resize((256, 256)),

    transforms.RandomResizedCrop(
        224,
        scale=(0.8, 1.0)
    ),

    transforms.RandomHorizontalFlip(),

    transforms.RandomRotation(10),

    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


vit_eval_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ============================================================
# 4. DATASETS
# ============================================================

train_dataset = datasets.ImageFolder(
    TRAIN_DIR,
    transform=vit_train_transform
)

val_dataset = datasets.ImageFolder(
    VAL_DIR,
    transform=vit_eval_transform
)

test_dataset = datasets.ImageFolder(
    TEST_DIR,
    transform=vit_eval_transform
)


print("\nClass mapping:")
print(train_dataset.class_to_idx)

print("\nDataset sizes:")
print("Train:", len(train_dataset))
print("Val  :", len(val_dataset))
print("Test :", len(test_dataset))


# ============================================================
# 5. DATALOADERS
# ============================================================

BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)


# ============================================================
# 6. LOAD PRETRAINED ViT-B/16
# ============================================================

print("\nLoading pretrained ViT-B/16...")

weights = models.ViT_B_16_Weights.DEFAULT

vit_model = models.vit_b_16(
    weights=weights
)


# ============================================================
# 7. REPLACE CLASSIFICATION HEAD
# ============================================================

NUM_CLASSES = 6

in_features = vit_model.heads.head.in_features

vit_model.heads.head = nn.Linear(
    in_features,
    NUM_CLASSES
)

vit_model = vit_model.to(device)


# ============================================================
# 8. MODEL INFORMATION
# ============================================================

total_params = sum(
    p.numel()
    for p in vit_model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in vit_model.parameters()
    if p.requires_grad
)

print("\nModel:")
print("Architecture     : ViT-B/16")
print("Pretrained       : ImageNet")
print("Classes          :", NUM_CLASSES)
print("Total parameters :", f"{total_params:,}")
print("Trainable        :", f"{trainable_params:,}")


# ============================================================
# 9. CLASS WEIGHTS
# ============================================================

# ImageFolder order:
#
# 0    -> 0
# MR   -> 1
# MRMS -> 2
# MS   -> 3
# R    -> 4
# S    -> 5

class_weights = torch.tensor(
    [
        4.4184,
        1.6036,
        0.7958,
        0.5031,
        2.5073,
        0.6637
    ],
    dtype=torch.float32,
    device=device
)


criterion = nn.CrossEntropyLoss(
    weight=class_weights
)


# ============================================================
# 10. DIFFERENT LEARNING RATES
# ============================================================

# Pretrained ViT backbone:
# small learning rate = 1e-5
#
# New classification head:
# larger learning rate = 1e-4

optimizer = optim.AdamW(
    [
        {
            "params": vit_model.conv_proj.parameters(),
            "lr": 1e-5
        },
        {
            "params": vit_model.encoder.parameters(),
            "lr": 1e-5
        },
        {
            "params": vit_model.class_token,
            "lr": 1e-5
        },
        {
            "params": vit_model.encoder.pos_embedding,
            "lr": 1e-5
        },
        {
            "params": vit_model.heads.parameters(),
            "lr": 1e-4
        }
    ],
    weight_decay=1e-4
)


# ============================================================
# 11. COSINE LR SCHEDULER
# ============================================================

NUM_EPOCHS = 10

scheduler = optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=NUM_EPOCHS,
    eta_min=1e-7
)


# ============================================================
# 12. MIXED PRECISION
# ============================================================

if device.type == "cuda":

    scaler = torch.amp.GradScaler("cuda")

else:

    scaler = None


# ============================================================
# 13. TRAINING CONFIG
# ============================================================

best_val_f1 = 0.0

CHECKPOINT_PATH = (
    "/content/best_yellow_rust_vit_b16_improved.pth"
)


# ============================================================
# 14. TRAINING
# ============================================================

print("\n")
print("=" * 75)
print("STARTING IMPROVED ViT-B/16 TRAINING")
print("=" * 75)


for epoch in range(NUM_EPOCHS):

    # ========================================================
    # TRAIN
    # ========================================================

    vit_model.train()

    running_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader:

        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )


        # ----------------------------------------------------
        # Forward + loss
        # ----------------------------------------------------

        if device.type == "cuda":

            with torch.autocast(
                device_type="cuda",
                dtype=torch.float16
            ):

                outputs = vit_model(images)

                loss = criterion(
                    outputs,
                    labels
                )

            scaler.scale(loss).backward()

            scaler.step(optimizer)

            scaler.update()

        else:

            outputs = vit_model(images)

            loss = criterion(
                outputs,
                labels
            )

            loss.backward()

            optimizer.step()


        # ----------------------------------------------------
        # Statistics
        # ----------------------------------------------------

        running_loss += (
            loss.item() * images.size(0)
        )

        predictions = outputs.argmax(
            dim=1
        )

        train_total += labels.size(0)

        train_correct += (
            predictions == labels
        ).sum().item()


    train_loss = (
        running_loss / train_total
    )

    train_acc = (
        train_correct / train_total
    )


    # ========================================================
    # VALIDATION
    # ========================================================

    vit_model.eval()

    val_loss_total = 0.0

    val_correct = 0
    val_total = 0

    all_val_labels = []
    all_val_predictions = []


    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )


            if device.type == "cuda":

                with torch.autocast(
                    device_type="cuda",
                    dtype=torch.float16
                ):

                    outputs = vit_model(images)

                    loss = criterion(
                        outputs,
                        labels
                    )

            else:

                outputs = vit_model(images)

                loss = criterion(
                    outputs,
                    labels
                )


            val_loss_total += (
                loss.item() * images.size(0)
            )

            predictions = outputs.argmax(
                dim=1
            )


            val_total += labels.size(0)

            val_correct += (
                predictions == labels
            ).sum().item()


            all_val_labels.extend(
                labels.cpu().numpy()
            )

            all_val_predictions.extend(
                predictions.cpu().numpy()
            )


    val_loss = (
        val_loss_total / val_total
    )

    val_acc = (
        val_correct / val_total
    )


    # ========================================================
    # VALIDATION MACRO F1
    # ========================================================

    val_macro_f1 = f1_score(
        all_val_labels,
        all_val_predictions,
        average="macro",
        zero_division=0
    )


    # ========================================================
    # CURRENT LEARNING RATE
    # ========================================================

    backbone_lr = optimizer.param_groups[0]["lr"]
    head_lr = optimizer.param_groups[-1]["lr"]


    # ========================================================
    # PRINT RESULTS
    # ========================================================

    print(
        f"Epoch [{epoch + 1}/{NUM_EPOCHS}] "
        f"| Train Loss: {train_loss:.4f} "
        f"| Train Acc: {train_acc:.4f} "
        f"| Val Loss: {val_loss:.4f} "
        f"| Val Acc: {val_acc:.4f} "
        f"| Val Macro-F1: {val_macro_f1:.4f}"
    )

    print(
        f"   LR → Backbone: {backbone_lr:.2e} "
        f"| Head: {head_lr:.2e}"
    )


    # ========================================================
    # SAVE BEST MODEL BY MACRO F1
    # ========================================================

    if val_macro_f1 > best_val_f1:

        best_val_f1 = val_macro_f1

        torch.save(
            vit_model.state_dict(),
            CHECKPOINT_PATH
        )

        print(
            f"   ✓ BEST MODEL SAVED "
            f"(Macro-F1: {val_macro_f1:.4f})"
        )


    # ========================================================
    # UPDATE LR
    # ========================================================

    scheduler.step()


# ============================================================
# 15. TRAINING COMPLETE
# ============================================================

print("\n")
print("=" * 75)
print("IMPROVED ViT TRAINING COMPLETE")
print("=" * 75)

print(
    f"Best Validation Macro-F1: "
    f"{best_val_f1:.4f}"
)

print(
    "Checkpoint:"
)

print(
    CHECKPOINT_PATH
)
```
